# 1. First-Order Linear Ordinary Differential Equation (ODE)

This is the foundational equation for linear systems. It is solved using an "integrating factor."

Visible Formula:
$$
\frac{dy}{dx} + P(x)y = Q(x)
$$

General Solution:
$$
y = e^{-\int P(x)dx} \left( \int Q(x)^{\int P(x)dx} dx + C \right)
$$

LaTeX
```
\frac{dy}{dx} + P(x)y = Q(x)

y = e^{-\int P(x)dx} \left( \int Q(x)e^{\int P(x)dx} dx + C \right)
```

AsciiMath
```
(dy)/(dx) + P(x)y = Q(x)

y = e^(-int P(x) dx) (int Q(x) e^(int P(x) dx) dx + C)
```

````python
import numpy as np
import matplotlib.pyplot as plt
from scipy.integrate import solve_ivp

# Define the ODE: dy/dx + 2y = e^(-x)  => dy/dx = e^(-x) - 2y
def ode_system(x, y):
    return np.exp(-x) - 2 * y

# Solve numerically
x_span = (0, 5)
y0 = [1.0] # Initial condition y(0) = 1
x_eval = np.linspace(0, 5, 200)
sol = solve_ivp(ode_system, x_span, y0, t_eval=x_eval)

# Exact analytical solution for comparison: y = (1/3)e^(-x) + (2/3)e^(-2x)
y_exact = (1/3)*np.exp(-x_eval) + (2/3)*np.exp(-2*x_eval)

# Plotting
plt.figure(figsize=(8, 5))
plt.plot(sol.t, sol.y[0], 'b--', label='Numerical Solution (solve_ivp)')
plt.plot(x_eval, y_exact, 'r-', linewidth=2, label='Exact Analytical Solution')
plt.title("First-Order Linear ODE: y' + 2y = e^(-x)")
plt.xlabel("x")
plt.ylabel("y")
plt.legend()
plt.grid(True)
plt.show()

"""
Graph Description:
The plot shows a curve starting at y = 1 when x = 0,
decaying exponentially towards 0 as x increases.
The numerical and exact solutions overlap perfectly.
"""
````
