# 1. First-Order Linear Ordinary Differential Equation (ODE)

This is the foundational equation for linear systems. It is solved using an "integrating factor."

Visible Formula:
$$
\frac{dy}{dx} + P(x)y = Q(x)
$$

General Solution:
$$
y = e^{-\int P(x)dx} \left( \int Q(x)^{\int P(x)dx} dx + C \right)
$$
