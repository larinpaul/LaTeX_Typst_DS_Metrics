# 1. Limit Definition of the Derivative

The foundational definition of the derivative, representing the instantaneous rate of change or the slope of teh tangent line.

Visible Formula:
$$
f'(x) = \lim_{h \to 0} \frac{f(x+h) - f(x)}{h}
$$

LaTeX Code:
```
f'(x) = \lim_{h \to 0} \frac{f(x+h) - f(x)}{h}
```

AsciiMath Code:
f'(x) = lim_(h->0) (f(x+h) - f(x))/h

````python
import sympy as sp

x, h = sp.symbols('x h')
f = sp.Function('f')

# Define the limit expression
limit_expr = (f(x + h) - f(x)) / h

# Compute the limit as h approaches 0
derivative = sp.limit(limit_expr, h, 0)
print("Limit Definition:", derivative)
````

# 2. Power Rule

The most basic and frequently used rule for differentiationg polynomial terms.

Visible Formula:
$$
\frac{d}{dx} [x^n] = n x^{n-1}
$$

LaTeX Code:
```
\frac{d}{dx} [x^n] = n x^{n-1}
```

AsciiMath Code:
```
d/dx [x^n] = n x^(n-1)
```


````python
import sympy as sp

x, n = sp.symbols('x n')
f = x**n

# Compute the derivative with respect to x
derivative = sp.diff(f, x)
print("Power Rule Result:", derivative)
````

# 3. Product Rule
Used to find the derivative of the product of two differentiable functions

Visible Formula:
$$
\frac{d}{dx} [f(x)g(x)] = f'(x)g(x) + f(x)g'(x)
$$

LaTeX Code:
```
\frac{d}{dx} [f(x)g(x)] = f'(x)g(x) + f(x)g'(x)
```

AsciiMath Code:
```
d/dx [f(x)g(x)] = f'(x)g(x) + f(x)g'(x)
```

````python
import sympy as sp

x = sp.symbols('x')
f = sp.Function('f')(x)
g = sp.Function('g')(x)

# Compute the derivative of the product
product_rule = sp.diff(f * g, x)
print("Product Rule Result:", product_rule)
````

# 4. Quotient Rule

Правило дифференцирования дроби (literally: "rule for differentiating a fraction") or правило производной частного (rule of the derivative of a quotient).
Spanish: la regla del cociente (or la regla de la derivada del cociente)

Used to find the derivative of the ratio of two differentiable functions.

Visible Formula:
$$
\frac{d}{dx} \left[ \frac{f(x)}{g(x)} \right] = \frac{f'(x)g(x) - f(x)g'(x)}{[g(x)]^2}
$$

LaTeX Code:
```
\frac{d}{dx} \left[ \frac{f(x)}{g(x)} \right] = \frac{f'(x)g(x) - f(x)g'(x)}{[g(x)]^2}
```

AsciiMath Code:
```
d/dx [f(x)/g(x)] = (f'(x)g(x) - f(x)g'(x))/(g(x))^2
```

````python
import sympy as sp

x = sp.symbols('x')
f = sp.Function('f')(x) # Chained function call, # writing func()() means you are calling a function, and that function returns another callable object (like a function or expression) that you immediately call with a second set of parentheses.
# In SymPy, you see this most often with laambdify, which converts a symbolic math expression into a standard, fast Python function
g = sp.Function('g')(x)

# Compute the derivative of the quotient
quotient_rule = sp.diff(f / g, x)
print("Quotient Rule Result:", sp.simplify(quotient_rule))
````

# 5. Chain Rule

Used to differentiate composite functions (a function of a function).

Visible Formula:
$$
\frac{d}{dx} [f(g(x))] = f'(g(x)) \cdot g'(x)
$$

LaTeX Code:
```
\frac{d}{dx} [f(g(x))] = f'(g(x)) \cdot g'(x)x
```

AsciiMath Code:
```
d/dx [f(g(x))] = f'(g(x)) * g'(x)
```

````python
import sympy as sp

x = sp.symbols('x')
g = sp.Function('g')(x)
f = sp.Function('f')(g)

# Compute the derivative of the compoite function
chain_rule = sp.diff(f, x)
print("Chain Rule Result:", chain_rule)
````

# 1. Numerical Differentiation (Finite Difference Method)

In data science, you often have discrete data points (e.g, stock prices, sensor readings) rather than a clean formula. We approximate the derivative using the Central Difference formula, which is more accurate than the basic forward difference.

Visible Formula:
$$
f'(x) \approx \frac{f(x+h) - f(x-h)}{2h}
$$

LaTeX Code:
```
f'(x) \approx \frac{f(x+h) - f(x-h)}{2h}
```

AsciiMath Code:
```
f'(x) approx (f(x+h) - f(x-h)) / (2h)
```

````python
import numpy as np

# Example: We have discrete data points for positions over time
t = np.array([0.0, 1.0, 2.0, 3.0, 4.0]) # Time in seconds
x = np.array([0.0, 2.1, 7.8, 18.2, 32.5]) # Position in meters (noisy data)

# NumPy's built-in gradient function uses second-order accurate central differences
# for interior points, and first-order for edges.
velocity = np.gradient(x, t)

print("Time points:     ", t)
print("Position data:   ", x)
print("Approx Velocity: ", np.round(velocity, 2))

````

Here is how Calculus 1 concepts scale up to handle millions of parameters in modern Data Science and Engineering using PyTorch (the same principles apply to TensorFlow/JAX).

The secret sauce is Automatic Differentiation (Autograd).
Unlike Symbolic Differentiation (which gets too slow with complex formulas) or Numerical Differentiation (which is inaccurate and slow for high dimensions), Autograd builds a computational graph during the forward pass and applies the Chain Rule perfectly and efficiently during the backward pass.


# 1. Automatic Differentiation (The Engine of Deep Learning)

Instead of writing derivative formulas by hand, we tell PyTorch to track operations on tensors. When we call .backward(), it automatically computes the exact derivatives using the chain rule.

Partial d - D parcial / De de Jacobi / Знак частной производной

Visible Formula:
$$
\frac{\partial L}{\partial w} = \frac{\partial L}{\partial y} \cdot \frac{\partial y}{\partial z} \cdot \frac{\partial z}{\partial w}
$$

LaTeX Code:
```
\frac{\partial L}{\partial w} = \frac{\partial L}{\partial y} \cdot \frac{\partial y}{\partial z} cdot \frac{\partial z}{\partial w}
```

AsciiMath Code:
```
(del L)/(del w) = (del L)/(del y) * (del y)/(del z) * (del z)/(de
```


````python
import torch

# 1. Define variables and tell PyTorch to track their gradients
w = torch.tensor(2.0, requires_grad=True)  # Weight
x = torch.tensor(3.0)                      # Input (no gradient needed)

# 2. Forward pass: Build the computational graph
# Let z = w * x, and y = z^2, and Loss L = (y - 10)^2
z = w * x
y = z**2
L = (y - 10.0)**2

# 3. Backward pass: Apply the Chain Rule automatically
L.backward()

# 4. Extract the computed derivative (dL/dw)
print(f"Computed gradient dL/dw: {w.grad.item():.4f}")

# Manual verification:
# dL/dy = 2*(y-10), dy/dz = 2*z, dz/dw = x
# At w=2, x=3: z=6, y=36, L=(36-10)^2 = 676
# dL/dw = 2*(36-10) * 2*(6) * 3 = 52 * 12 * 3 = 1872.0
print(f"Manual verification: {1872.0}")
````

# 1. The Derivative: Slope of the Tangent Line

The derivative at a specific point is simply the slope of the line that just "touches" the curve at that exact point. It tells you the instantaneous rate of change.

Visible Formula:
$$
f'(x_0) = \lim_{h \to 0} \frac{f(x_0+h) - f(x_0)}{h} = m_{\text{tangent}}
$$

LaTeX Code:
```
f'(x_0) = \lim_{h \to 0} \frac{f(x_0+h) - f(x_0)}{h} = m_{\text{tangent}}
```

AsciiMath Code:
```
f'(x_0) = lim_(h->0) (f(x_0+h) - f(x_0))/h = m_"tangent"
```


````python
import numpy as np
import matplotlib.pyplot as plt

# Define the function and its derivative
x = np.linspace(-1, 5, 200)
y = x**2  # f(x) = x^2

# Point of tangency
x0 = 2.0
y0 = x0**2
slope = 2 * x0  # f'(x) = 2x, at x=2, slope=4

# Tangent line equation: y - y0 = m(x - x0)
x_tan = np.linspace(0, 4, 100)
y_tan = y0 + slope * (x_tan - x0)

plt.figure(figsize=(8, 5))
plt.plot(x, y, 'b-', linewidth=2, label='f(x) = x²')
plt.plot(x_tan, y_tan, 'r-', linewidth=2, label=f'Tangent line (slope = {slope})')
plt.plot(x0, y0, 'ro', markersize=8, label=f'Tangent point ({x0}, {y0})')

plt.title('The Derivative as the Slope of a Tangent Line', fontsize=14)
plt.xlabel('x', fontsize=12); plt.ylabel('f(x)', fontsize=12)
plt.legend(); plt.grid(True, alpha=0.6)
plt.show()
````

## Here is the ultimate breakdown of how Custom Autograd works under the hood.

To make this concrete, we will build a custom activation function: the Softplus function,
defined as f(x) = ln(1 + e^x). It is a smooth, diferentiable approximation of the ReLU function.
Its derivative is beautifully simple: the Sigmoid function, f'(x) = (e^x)/(1+e^x).

We will implement this in PyTorch (using its Autograd engine), buidl the exact same mechanics from scratch in Numpy, and then graph it to prove the math matches the geometry.

The Custom Function & Its Derivative

Visible Formula:
$$
f(x) = \ln(1 + e^x) \implies f'(x) = \frac{e^x}{1 + e^x}
$$

LaTeX Code:
```
f(x) = \ln(1 + e^x) \implies f'(x) = \frac{e^x}{1 + e^x}
```

AsciiMath Code:
```
f(x) = ln(1 + e^x0 implies f'(x) = e^x / (1 + e^x)
```

````python
# 1. PyTorch Custom Autograd Implementation

"""
In PyTorch, you create a custom operation by subclassing torch.autograd.Function.
You must define two static methods: forward (which saves variables for later) and backward (which uses the chain rule to pass gradient backward).
"""

import torch

class CustomSoftplus(torch.autograd.Function):
    @staticmethod
    def forward(ctx, x):
        # 1. Compute the forward pass
        # We use log1p(exp(x)) for numerical stability to prevent overflow
        result = torch.log1p(torch.exp(x))

        # 2. Save 'x' in the context object for the backward pass
        ctx.save_for_backward(x)

        return result

    @staticmethod
    def backward(ctx, grad_output):
        # 1. Retrieve the saved 'x' from the forward pass
        x, = ctx.saved_tensors

        # 2. Compute the local derivative: f'(x) = e^x / (1 + e^x)  (which is Sigmoid)
        # We use torch.sigmoid for numerical stability
        local_grad = torch.sigmoid(x)

        # 3. Apply the Chain Rule: multiply upstream gradient by local gradient
        grad_input = grad_output * local_grad

        return grad_input

# --- Testing the Custom Autograd ---
x = torch.tensor([0.0, 2.0, -2.0], requires_grad=True)

# Forward pass
y = CustomSoftplus.apply(x)
print("Forward Output (Softplus):", y)

# Backward pass (assume upstream gradient is [1.0, 1.0, 1.0])
y.backward(torch.ones_like(x))
print("Backward Output (Gradients):", x.grad)

# Verification against PyTorch's built-in softplus
y_builtin = torch.nn.functional.softplus(x)
y_builtin.backward(torch.ones_like(x))
print("Built-in Gradients (Match?):", torch.allclose(x.grad, torch.tensor([0.5000, 0.8808, 0.1192], dtype=torch.float32)))
````

````python
# 2. Pure NumPy Analogue (The "Under the Hood" Mechanics)

"""
PyTorch's Autograd is essentially a sophisticated version of this NumPy class.
It tracks operations, saves the forward state, and manually applies the chain rle during the backward step.
"""

import numpy as np

class NumpySoftplus:
    def __init__(self):
        self.x = None # To store the forward state

    def forward(self, x):
        # 1. Save state for backward pass
        self.x = x

        # 2. Compute forward: log(1 + exp(x))
        # np.log1p is numerically stable for small x
        return np.log1p(np.exp(x))

    def backward(self, grad_output):
        # 1. Compute local derivative: sigmoid(x)
        # We use a stable sigmoid implementation: 1 / (1 + exp(-x))
        local_grad = 1.0 / (1.0 + np.exp(-self.x))

        # 2. Apply Chain Rule
        return grad_output * local_grad

# --- Testing the NumPy Analogue ---
x_np = np.array([0.0, 2.0, -2.0])
grad_upstream = np.array([1.0, 1.0, 1.0])

# Instantiate and run forward
op = NumpySoftplus()
y_np = op.forward(x_np)
print("\nNumPy Forward Output:", np.round(y_np, 4))

# Run backward
grad_np = op.backward(grad_upstream)
print("NumPy Backward Output (Gradients):", np.round(grad_np, 4))
````

````python
# 3. Graphical Proof: Function, Derivative, and Tangent Line

"""
A formula is just abstract until you see it.
This graph plots the Softplus function (blue),
its derivative/Sigmoid (orange), and a tangent line at x = 0.

Notice that at x = 0, the derivative curve is exactly at 0.5. Consequently, the tangent line drawn on the Softplus curve at x = 0 has a slope of exactly 0.5. This is the geometric proof that out Autorad backward pass is correct.
"""
import numpy as np
import matplotlib.pyplot as plt

# 1. Generate data
x = np.linspace(-4, 4, 400)
y_func = np.log1p(np.exp(x))          # f(x) = Softplus
y_deriv = 1 / (1 + np.exp(-x))        # f'(x) = Sigmoid

# 2. Tangent line at x = 0
x0 = 0.0
y0 = np.log1p(np.exp(x0))             # f(0) = ln(2) ≈ 0.693
slope = 1 / (1 + np.exp(-x0))         # f'(0) = 0.5

# Equation of tangent line: y = m(x - x0) + y0
x_tan = np.linspace(-2, 2, 100)
y_tan = slope * (x_tan - x0) + y0

# 3. Plotting
fig, ax1 = plt.subplots(figsize=(10, 6))

# Plot the Function and Tangent Line (Left Y-axis)
ax1.plot(x, y_func, 'b-', linewidth=2.5, label='f(x) = Softplus')
ax1.plot(x_tan, y_tan, 'r--', linewidth=2, label=f'Tangent at x=0 (Slope = {slope})')
ax1.plot(x0, y0, 'ro', markersize=8, label='Point (0, ln(2))')
ax1.set_xlabel('x', fontsize=12)
ax1.set_ylabel('f(x)  (Function Value)', color='b', fontsize=12)
ax1.tick_params(axis='y', labelcolor='b')
ax1.grid(True, alpha=0.3)

# Plot the Derivative (Right Y-axis)
ax2 = ax1.twinx()
ax2.plot(x, y_deriv, 'g-', linewidth=2.5, label="f'(x) = Sigmoid (Derivative)")
ax2.plot(x0, slope, 'go', markersize=8, label="f'(0) = 0.5")
ax2.set_ylabel("f'(x)  (Slope / Gradient)", color='g', fontsize=12)
ax2.tick_params(axis='y', labelcolor='g')

# Combine legends
lines1, labels1 = ax1.get_legend_handles_labels()
lines2, labels2 = ax2.get_legend_handles_labels()
ax1.legend(lines1 + lines2, labels1 + labels2, loc='upper left', fontsize=11)

plt.title('Custom Autograd Verified: Softplus Function and its Derivative', fontsize=14)
plt.tight_layout()
plt.show()
````
