```python
import sympy as sp

x, u = sp.symbols('x u')
# Example: Integrate 2x * cos(x^2)
integrand = 2 * x * sp.cos(x**2)

# SymPy automatically applies u-substitution internally
result = sp.integrate(integrand, x)
print("U-Substitution Result:", result)

# Manual verification: let u = x^2, du = 2x dx
manual_result = sp.integrate(sp.cos(u), u).subs(u, x**2)
print("Manual U-Sub Verification:", manual_result)
```
