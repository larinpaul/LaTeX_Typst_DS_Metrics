#### Linear algebra is the mathematics of data, spaces, and linear transformations. Here are the most fundamental concepts and formulas

#### 1. Vector Norm (Magnitude / Length)

$$
\|\mathbf{a}\| = \sqrt{\sum_{i=1}^{n} a_i^2} = \sqrt{\mathbf{a} \cdot \mathbf{a}}
$$

---



```
LaTeX Code:
\|\mathbf{a}\| = \sqrt{\sum_{i=1}^{n} a_i^2} = \sqrt{\mathbf{a} \cdot \mathbf{a}}

AsciiMath Code:
||vec(a)|| = sqrt(sum_(i=1)^n a_i^2) = sqrt(vec(a) cdot vec(a))

```


````python
import numpy as np

a = np.array([3, 4])
# Calculate the L2 norm (Euclidian length)
norm_a = np.linalg.norm(a)
print(norm_a) # Output: 5.0
````

### 2. The Dot Prodct (Inner Product)

Explanation: The dot product multiplies two vectors together to yield a single scalar (number). Intuitively, it measures how much two vectors point in the same direction. If the dot product is 0, the vectors are perfectly perpendicular (orthogonal). It is the foundation of projections and similrity measures (like cosine simialrity).

$$
\mathbf{a} \cdot \mathbf{b} = \sum_{i=1}^{n} a_i b_i = \mathbf{a}^T \mathbf{b}
$$

```
LaTeX Code:
\mathbf{a} \cdot \mathbf{b} = \sum_{i=1}^{n} a_i b_i = \mathbf{a}^T \mathbf{b}

AsciiMath Code:
vec(a) cdot vec(b) = sum_(i=1)^n a_i b_i = vec(a)^T vec(b)

```

````python
import numpy as np

a = np.array([1, 2, 3])
b = np.array([4, 5, 6])

# Calculate dot product
dot_product = np.dot(a, b)
# Alternatively, you can use the @ operator: a @ b
print(dot_product) # Output: 32 (1*4 + 2*5 + 3*6)
````

# 3. Matrix Multiplication

Explanation: Matrix multiplication is how we combine linear transformations (like rotating, scaling, or shearing space). TO get the element in the i-th row and j-th column of the resulting matrix, you take the dot product of the i0th row of the first matrix and the j-th column of the second matrix. Note: Matrix multiplication is *not* communtative (AB != BA).

1. Visible Formula:
$$(AB)_{ij} = \sum_{k=1}^{n} A_{ik} B_{kj}$$

2. LaTeX Code:
```
(AB)_{ij} = \sum_{k=1}^{n} A_{ik} B_{kj}
```

3. AciiMath Code:
(AB)_(ij) = sum_(k=1)^n A_(ij) B_(kj)
