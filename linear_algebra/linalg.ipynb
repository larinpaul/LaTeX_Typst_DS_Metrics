#### Linear algebra is the mathematics of data, spaces, and linear transformations. Here are the most fundamental concepts and formulas

#### 1. Vector Norm (Magnitude / Length)

$$
\|\mathbf{a}\| = \sqrt{\sum_{i=1}^{n} a_i^2} = \sqrt{\mathbf{a} \cdot \mathbf{a}}
$$

---



```
LaTeX Code:
\|\mathbf{a}\| = \sqrt{\sum_{i=1}^{n} a_i^2} = \sqrt{\mathbf{a} \cdot \mathbf{a}}

AsciiMath Code:
||vec(a)|| = sqrt(sum_(i=1)^n a_i^2) = sqrt(vec(a) cdot vec(a))

```


````python
import numpy as np

a = np.array([3, 4])
# Calculate the L2 norm (Euclidian length)
norm_a = np.linalg.norm(a)
print(norm_a) # Output: 5.0
````

### 2. The Dot Prodct (Inner Product)

Explanation: The dot product multiplies two vectors together to yield a single scalar (number). Intuitively, it measures how much two vectors point in the same direction. If the dot product is 0, the vectors are perfectly perpendicular (orthogonal). It is the foundation of projections and similrity measures (like cosine simialrity).

$$
\mathbf{a} \cdot \mathbf{b} = \sum_{i=1}^{n} a_i b_i = \mathbf{a}^T \mathbf{b}
$$

```
LaTeX Code:
\mathbf{a} \cdot \mathbf{b} = \sum_{i=1}^{n} a_i b_i = \mathbf{a}^T \mathbf{b}

AsciiMath Code:
vec(a) cdot vec(b) = sum_(i=1)^n a_i b_i = vec(a)^T vec(b)

```

````python
import numpy as np

a = np.array([1, 2, 3])
b = np.array([4, 5, 6])

# Calculate dot product
dot_product = np.dot(a, b)
# Alternatively, you can use the @ operator: a @ b
print(dot_product) # Output: 32 (1*4 + 2*5 + 3*6)
````

# 4. The Determinant

Explanation: The determinant is a scalar value that can be computed
from a square matrix. Geometrically, it represets
the scaling factor of the area (or volume) when a transformation is applied.

If det(A) = 0, the matrix squishes space into a lower dimension
(it's "singular" and cannot be inverted).
If det(A) < 0, the transformation flip the orientation of the space.

1. Visible Formula (for a 2x2 matrix):
$$\det(A) = \begin{vmatrix} a & b \\ c & d \end{vmatrix} = ad - bc$$

2. LaTeX Code:
```
\det(A) = \begin{vmatrix} a & b \\ c & d \end{vmatrix} = ad - bc
```

3. AsciiMath Code:
```
det(A) = |[[a,b],[c,d]]| = ad - bc
```

````python
# 4. Python Code:
import numpy as np

A = np.array([[1, 2],
              [3, 4]])

# Calculate determinant
det_A = np.linalg.det(A)
print(det_A) # Output: -2.0 (since 1*4 - 2*3 = -2)
````

# 5. Matrix Inverse

Explanation: The inverse of a matrix A is another matrix A^-1 that completely "undoes" the transformation of A. When you multiply a matrix by its inverse, you get the Identity matrix (I), which is the matrix equivalent of the number 1. A matrix only has the inverse of its determinant is not zero.

1. Visible Formula:
$$
A A^{-1} = A^{-1} A = I
$$

2. LaTeX Code:
```
A A^{-1} = A^{-1} A = I
```

3. AsciiMath Code:
```
A A^(-1) = A^(-1) A = I
```


````python
# 4. Python Code:
import numpy as np

A = np.array([[1, 2],
              [3, 4]])

# Calculate inverse
A_inv = np.linalg.inv(A)
print(A_inv)
# Output: [[-2.    1. ]
#          [ 1.4  -0.5]]

# Verify: A @ A_inv should equal the Identity matrix
index_check = np.dot(A, A_inv)
````
