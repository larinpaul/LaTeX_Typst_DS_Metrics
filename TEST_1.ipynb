### Математика узла Дерева Решений (в Jupyter)

В узле находится **4** объекта класса (+) и **6** объектов класса (-)

Формула Энтропии Шеннона:
$$ H(X) = - \sum_{i=1}^{n} p_i \log_2 p_i $$
H(X) = - \sum_{i=1}^{n} p_i \log_2 p_i

**Пример расчета:**
$$ H(X) = - \left( \frac{4}{10} \log_2 \frac{4}{10} + \frac{6}{10} \right) \approx 0.971 $$
H(X) = - \left( \frac{4}{10} \log_2 \frac{4}{10} + \frac{6}{10} \right) \approx 0.971

** Высокая энтропия:** данные сильно перемешаны. Узел требует разделения!


````python
# Вариант 2. Интерактивный расчет на Python с выводом в LaTeX (в Code-ячейке) # IPython.display

import math
from IPython.display import display, Markdown

# --- 1. Исходные данные (комментации на Python через #) ---
pos = 4
neg = 6
total = pos + neg

# --- 2. Расчет энтропии
if total == 0:
    entropy = 0.0
else:
    p1 = pos / total
    p2 = neg / total

    # Считаем логарифмы с проверкой на 0
    log_p1 = math.log2(p1) if p1 > 0 else 0
    log_p2 = math.log2(p2) if p2 > 0 else 0

    entropy = - (p1 * log_p1 + p2 * log_p2)

# --- 3. Логика проверки состояния ---
if entropy > 0.9:
    status = "<span style='color:red;>** Высокая энтропия:** данные сильно перемешаны. Требуется разделение!</span>'"
else:
    status = "<span style='color:green;'>** Низкая энтропия: узел относительно чистый.**</span>"

# --- 4. Красивый вывод с использованием LaTeX-формул ---
output = rf"""
### Динамический расчет узла в Jupyter
В узле находится **{pos}** объектов (+) и **{neg}** объектов (-). Total = {total}.

Формула:
$$ H(X) = - \sum_{{i=1}}^{{n}} p_i \log_2 p_1 $$

Текущая энтропия:
$$\Large H(X) = {entropy:.3f}$$

{status}
"""

display(Markdown(output))

````

````python
# 2. Plotting Math Functions (Python Code Cell)

# Here is the clean matplotlib code to generate the continuous Shannon Entropy curve.
# It maps how uncertainty reaches its maximum at p = 0.5

import numpy as np
import matplotlib.pyplot as plt

# Generate probability values from 0.01 to 0.99 (avoiing 0 and 1 to prevent log(0) errors)
p = np.linspace(0.01, 0.99, 100)

# Binary Shannon Entropy formula: H(p) = -p*log2(p) - (1-p)*log2(1-p)
h = -p * np.log2(p) - (1 - p) * np.log2(1 - p)

plt.figure(figsize=(6, 4))
plt.plot(p, h, color='#2b6cb0', linewidth=2.5, label='$H(p)$')
plt.title('Shannon Entropy for Binary Classification', fontsize=12, fontweight='bold')
plt.xlabel('Probability ($p$)', fontsize=10)
plt.ylabel('Entropy $H(p$', fontsize=10)
plt.grid(True, linestyle='--', alpha=0.6)

# Highlight maximum uncertainty point
plt.axvline(x=0.5, color='red', linestyle=':', label='Max Uncertainty ($p=0.5$)')
plt.legend()
plt.show()


````
