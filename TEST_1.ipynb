### Математика узла Дерева Решений (в Jupyter)

В узле находится **4** объекта класса (+) и **6** объектов класса (-)

Формула Энтропии Шеннона:
$$ H(X) = - \sum_{i=1}^{n} p_i \log_2 p_i $$
H(X) = - \sum_{i=1}^{n} p_i \log_2 p_i

**Пример расчета:**
$$ H(X) = - \left( \frac{4}{10} \log_2 \frac{4}{10} + \frac{6}{10} \right) \approx 0.971 $$
H(X) = - \left( \frac{4}{10} \log_2 \frac{4}{10} + \frac{6}{10} \right) \approx 0.971

** Высокая энтропия:** данные сильно перемешаны. Узел требует разделения!


````python
# Вариант 2. Интерактивный расчет на Python с выводом в LaTeX (в Code-ячейке) # IPython.display

import math
from IPython.display import display, Markdown

# --- 1. Исходные данные (комментации на Python через #) ---
pos = 4
neg = 6
total = pos + neg

# --- 2. Расчет энтропии
if total == 0:
    entropy = 0.0
else:
    p1 = pos / total
    p2 = neg / total

    # Считаем логарифмы с проверкой на 0
    log_p1 = math.log2(p1) if p1 > 0 else 0
    log_p2 = math.log2(p2) if p2 > 0 else 0

    entropy = - (p1 * log_p1 + p2 * log_p2)

# --- 3. Логика проверки состояния ---
if entropy > 0.9:
    status = "<span style='color:red;>** Высокая энтропия:** данные сильно перемешаны. Требуется разделение!</span>'"
else:
    status = "<span style='color:green;'>** Низкая энтропия: узел относительно чистый.**</span>"

# --- 4. Красивый вывод с использованием LaTeX-формул ---
output = rf"""
### Динамический расчет узла в Jupyter
В узле находится **{pos}** объектов (+) и **{neg}** объектов (-). Total = {total}.

Формула:
$$ H(X) = - \sum_{{i=1}}^{{n}} p_i \log_2 p_1 $$

Текущая энтропия:
$$\Large H(X) = {entropy:.3f}$$

{status}
"""

display(Markdown(output))

````

````python
# 2. Plotting Math Functions (Python Code Cell)

# Here is the clean matplotlib code to generate the continuous Shannon Entropy curve.
# It maps how uncertainty reaches its maximum at p = 0.5

import numpy as np
import matplotlib.pyplot as plt

# Generate probability values from 0.01 to 0.99 (avoiing 0 and 1 to prevent log(0) errors)
p = np.linspace(0.01, 0.99, 100)

# Binary Shannon Entropy formula: H(p) = -p*log2(p) - (1-p)*log2(1-p)
h = -p * np.log2(p) - (1 - p) * np.log2(1 - p)

plt.figure(figsize=(6, 4))
plt.plot(p, h, color='#2b6cb0', linewidth=2.5, label='$H(p)$')
plt.title('Shannon Entropy for Binary Classification', fontsize=12, fontweight='bold')
plt.xlabel('Probability ($p$)', fontsize=10)
plt.ylabel('Entropy $H(p$', fontsize=10)
plt.grid(True, linestyle='--', alpha=0.6)

# Highlight maximum uncertainty point
plt.axvline(x=0.5, color='red', linestyle=':', label='Max Uncertainty ($p=0.5$)')
plt.legend()
plt.show()


````

````python
# 3. Grapth Theory (Python Code Cell)
# In mathematics, a graph consists of vertices connected by edges.
# Torender network layouts and directed/undirected graphs, we use Python's networkx library.
import networkx as nx
import matplotlib.pyplot as plt

# Create a Directed Graph object
G = nx.DiGraph()

# Add edges representing relationships between vertices
G.add_edges_from([('v1', 'v2'), ('v1', 'v3'), ('v2', 'v4'), ('v3', 'v4'), ('v4', 'v5')])

# Spring layout distributes nodes mathematically based on force-directed algorithms
pos = nx.spring_layout(G, seed=42)

plt.figure(figsize=(5, 3))
nx.draw(G, pos, with_labels=True, node_color='#cbd5e1', node_size=700,
        font_weight='bold', font_size=10, edge_color='#64748b',
        arrows=True, arrowsize=18, width=2)
plt.title("Graph Theory: Directed Graph (Acyclic)", fontsize=12, fontweight='bold')
plt.show()
````

````python
# 4. Decision Trees (Python Code Cell)
# To render a clear hierarchical decision tree rather than a tangled web,
# we can hardcode spatial (x,y) coordinates into a networkx dictionary.
# This simulates a top-down tree generation process.

import networkx as nx
import matplotlib.pyplot as plt

T = nx.DiGraph()

# Tree edges: Parent Node -> Child Node
T.add_edges_from([
    ('Outlook?', 'Sunny'), ('Outlook?', 'Rainy'),
    ('Sunny', 'Humidity?'), ('Humidity?', 'High: No Play'), ('Humidity?', 'Normal: Play'),
    ('Rainy', 'Wind?'), ('Wind?', 'Strong: No Play'), ('Wind?', 'Weak: Play')
])

# Precise (X, Y) coordinates mapping to force a hierarchical tree look
pos = {
    'Outlook?': (0, 2),
    'Sunny': (-1.5, 1), 'Rainy': (1.5, 1),
    'Humidity?': (-1.5, 0), 'Wind?': (1.5, 0),
    'High: No Play': (-2.2, -1), 'Normal: Play': (-0.8, -1),
    'Strong: No Play': (0.8, -1), 'Weak: Play': (2.2, -1)
}

plt.figure(figsize=(10, 5))

# Draw rectangular box nodes for decisions ('s' stsnds for squares)
nx.draw(T, pos, with_labels=True, node_size=3200, node_color='#f1f5f9',
        font_size=9, font_weight='bold', edge_color='#94a3b8',
        arrows=True, arrowsize=15, node_shape='s',
        bbox=dict(facecolor='none', edgecolor='#cbd5e1',boxstyle='round,pad=0.3'))

plt.title("Decision Tree Hierarchical Structure", fontsize=14, fontweight='bold')
plt.show()
````
